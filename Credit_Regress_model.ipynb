```python
sel_df=df[['Income','Rating','Student_Yes','Balance']]
sel_df.head()
```

```python
import matplotlib.pyplot as plt
```

```python
fig,ax=plt.subplots(ncols=2,figsize=(15,8))
ax[0].scatter(sel_df['Income'],sel_df['Balance'])
ax[0].set_title('Income versus Balance')
ax[0].set_xlabel('Income')
ax[0].set_ylabel('Balance')
ax[1].scatter(sel_df['Rating'],sel_df['Balance'])
ax[1].set_title('Rating versus Balance')
ax[1].set_xlabel('Rating')
ax[1].set_ylabel('Balance')
fig.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
```

```python
y=sel_df['Balance']
x=sel_df.drop(columns=['Balance'])
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=7)
```

```python
model=LinearRegression()
```

```python
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
mse=mean_squared_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)*100
```

```python
print(f"The mean squared error of the model is {mse:.2f}")
print(f"The models prediction that resulted from the features is {r2:.2f}")
```

```python
print(f"The intercept of the model's line is {model.intercept_}")
print(f"The coefficients of the model are \nIncome: {model.coef_[0]:.2f}  \nRating: {model.coef_[1]:.2f}  \nEducation: {model.coef_[2]:.2f} ")
```

```python
rmse=np.sqrt(mse)
print(f"The off factor scale of the prediction is {rmse:.2f}")
```

```python
#Lets check on scaling
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
```

```python
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.fit_transform(x_test)
```

```python
s_model=LinearRegression()
```

```python
s_model.fit(x_train_scaled,y_train)
```

```python
y_scaled_pred=s_model.predict(x_test_scaled)
```

```python
s_mse=mean_squared_error(y_test,y_scaled_pred)
s_rmse=np.sqrt(s_mse)
s_r2=r2_score(y_test,y_scaled_pred)*100
```

```python
print(f"The mean squared error of the model is {s_mse:.2f}")
print(f"The models prediction that resulted from the features is {s_r2:.2f}")
print(f"The off factor scale of the prediction is {s_rmse:.2f}")
```

```python
print(f"The intercept of the model's line is {s_model.intercept_}")
print(f"The coefficients of the model are \nIncome: {s_model.coef_[0]:.2f}  \nRating: {s_model.coef_[1]:.2f}  \nEducation: {s_model.coef_[2]:.2f} ")
```
