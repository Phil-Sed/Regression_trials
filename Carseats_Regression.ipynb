```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("Carseats.csv")
df.head()
```

```python
df.info()
```

```python
df=pd.get_dummies(df,columns=['ShelveLoc','Urban','US'],drop_first=True,dtype=int)
df.head()
```

```python
import seaborn as sns
```

```python
plt.figure(figsize=(15,5))
sns.heatmap(df.corr(),annot=True)
```

```python
df_sel=df[['Income','Advertising','Price','Age','ShelveLoc_Medium','US_Yes','Sales']]
df_sel.head()
```

```python
X=df_sel.drop(columns=['Sales'])
Y=df_sel['Sales']
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score,mean_squared_error
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=7)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
```

```python
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.fit_transform(X_test)
```

```python
model=LinearRegression()
```

```python
model.fit(X_train_scaled,Y_train)
```

```python
y_pred=model.predict(X_test_scaled)
```

```python
mse=mean_squared_error(Y_test,y_pred)
rmse=np.sqrt(mse)
r2=r2_score(Y_test,y_pred)*100
```

```python
print(f"The MSE of the model is {mse:.2f}")
print(f"The RMSE of the model is {rmse:.2f}")
print(f"The R2 Score of the model is {r2:.2f}")
print(f"The intercept of the model is {model.intercept_}")
print(f"The coefficients of the model are {model.coef_}")
```

```python
from sklearn.linear_model import Lasso, Ridge
```

```python
ridge_model= Ridge(alpha=10.0)
ridge_model.fit(X_train_scaled,Y_train)
print(f"Ridge Coefficients: {ridge_model.coef_}")
```

```python
lasso_model = Lasso(alpha=10.0)
lasso_model.fit(X_train_scaled,Y_train)
print(f"Lasso Coefficients: {lasso_model.coef_}")
```

```python
y_p=ridge_model.predict(X_test_scaled)
```

```python
r2=r2_score(Y_test,y_p)*100
print(f"{r2:.2f}")
```
