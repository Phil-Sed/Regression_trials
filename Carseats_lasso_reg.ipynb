```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df=pd.read_csv('Carseats.csv')
df.head()
```

```python
df=pd.get_dummies(df,columns=['ShelveLoc','Urban','US'],drop_first=True,dtype=int)
df.head()
```

```python
X=df.drop(columns=['Sales'])
Y=df['Sales']
```

```python
X.head()
```

```python
from sklearn.linear_model import LassoCV
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score,mean_squared_error
from sklearn.preprocessing import StandardScaler
```

```python
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2)
```

```python
scaler=StandardScaler()
```

```python
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.fit_transform(x_test)
```

```python
alpha_space=np.logspace(-3,4,200)
```

```python
lassocv=LassoCV(alphas=alpha_space,cv=5,random_state=7)
lassocv.fit(x_train_scaled,y_train)
```

```python
y_pred=lassocv.predict(x_test_scaled)
```

```python
r2_score(y_test,y_pred)*100
```

```python
rmse=np.sqrt(mean_squared_error(y_test,y_pred))
```

```python
print(rmse)
```

```python
print(y_test[:10],y_pred[:10])
```

```python
from sklearn.linear_model import RidgeCV
```

```python
ridgecv=RidgeCV(alphas=alpha_space,scoring='r2',cv=5)
ridgecv.fit(x_train_scaled,y_train)
```

```python
pred=ridgecv.predict(x_test_scaled)
r2_score(y_test,pred)*100
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model=LinearRegression()
```

```python
model.fit(x_train_scaled,y_train)
```

```python
p=model.predict(x_test_scaled)
r2_score(y_test,p)*100
```
