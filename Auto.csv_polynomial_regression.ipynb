```python
fig,ax=plt.subplots(ncols=2,nrows=1,figsize=(10,5))
ax[0].scatter(df['horsepower'],df['mpg'])
ax[0].set_title('Horsepower versus mpg')
ax[0].set_xlabel('Horsepower')
ax[0].set_ylabel('mpg')
ax[1].scatter(df['acceleration'],df['mpg'])
ax[1].set_title('acceleration versus mpg')
ax[1].set_xlabel('acceleration')
ax[1].set_ylabel('mpg')
fig.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.metrics import r2_score
```

```python
X=df.drop(columns=['mpg'])
y=df['mpg']
```

```python
for degree in [1,2,3]:
    model=make_pipeline(PolynomialFeatures(degree=degree, include_bias=False),LinearRegression())
```

```python
model.fit(X,y)
```

```python
y_pred=model.predict(X)
```

```python
r2_score(y,y_pred)*100
```
